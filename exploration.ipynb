# SupplySense AI — Exploratory Data Analysis
Quick sanity-check plots over the generated synthetic dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

suppliers = pd.read_csv('../data/suppliers.csv')
orders = pd.read_csv('../data/orders.csv')
inventory = pd.read_csv('../data/inventory.csv')
suppliers.head()
```

```python
suppliers['risk_level'].value_counts().plot(kind='bar', title='Supplier Risk Distribution')
plt.show()
```

```python
orders['status'].value_counts().plot(kind='pie', autopct='%1.0f%%', title='Order Status')
plt.ylabel('')
plt.show()
```

```python
completed = orders[orders['status'] != 'In Transit'].copy()
completed['order_date'] = pd.to_datetime(completed['order_date'])
completed['month'] = completed['order_date'].dt.to_period('M').astype(str)
delay_rate = completed.groupby('month').apply(lambda g: (g['status']=='Delayed').mean(), include_groups=False)
delay_rate.plot(title='Monthly Delay Rate', figsize=(10,4))
plt.ylabel('Delay rate')
plt.show()
```

```python
inventory.groupby('category')[['current_stock','reorder_level']].sum().plot(kind='bar', title='Stock vs Reorder by Category', figsize=(10,4))
plt.show()
```
